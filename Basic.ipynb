```python
# Build a simple calculator for addition, subtraction, multiplication, division
def get_calculation():
    """ Take user input and validate it
    """
    
    calculation = input("Enter calculation with format { num op num } (eg. 1 + 1):\n").split()
    if len(calculation) != 3:
        print("\nInvalid calculation! Correct format is { num op num }")
        return None
    if calculation[1] not in ('+', '-', '*', '/'):
        print(f"\nInvalid operator '{calculation[1]}'!")
        return None

    num1 = int(calculation[0])
    op = calculation[1]
    num2 = int(calculation[2])

    return num1, op, num2

def calculate(calculation):
    num1, op, num2 = calculation

    match op:
        case '+':
            result = num1 + num2
        case '-':
            result = num1 - num2
        case '*':
            result = num1 * num2
        case '/':
            if num2 == 0:
                print("Cannot divide by zero!")
                return None
            else:
                result = num1 / num2
        case _:
            return None

    return result

calculation = get_calculation()
if calculation is not None:
    result = calculate(calculation)
    if result is not None:
        print(f"\nResult = {result}")
```

```python
# find factorial of a given number

num = int(input("Enter a number: "))

def factorial(num):
    if num == 0:
        return 1
    elif num <= 0:
        print("Cannot calculate factorial of a negative number!")
        return None

    factorial = 1
    for i in range(1, num + 1):
        factorial *= i

    return factorial

fac = factorial(num)
if fac is not None:
    print(f"{num}! = {fac}")
```

```python
# check if number is prime or not
num = int(input("Enter a number: "))

def check_prime(num):
    if num <= 1:
        print(f"{num} is neither prime nor composite")
        return None

    for i in range(2, num // 2):
        if num % i == 0:
            return False

    return True

is_prime = check_prime(num)

if is_prime is not None:
    if is_prime:
        print(f"{num} is a prime number.")
    else:
        print(f"{num} is not a prime number.")
```

```python
# print prime numbers between 10 and 99
print("Prime number between 10 and 99 are as follows:")

for i in range(10, 100):
    if check_prime(i):
        print(i, end=" ")
```

```python
# determine whether a given year is a leap year or not.

# a year is a leap year if it is divisible by 4
# if it is divisible by 100, it is not a leap year unless it is also divisible by 400

year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("It is a leap year")
else:
    print("Not a leap year")
```

```python
# In certain university, grades are assigned to the marks obtained in respective subjects. Grades are assigned as given below.Write a python program to accept Marks in Math and display respective grade as an output using if…elif…elif..else statements (Using multiple conditions).
# Hint : Relational operators: <, >, <=, >=, ==, !=
# Marks obtained    Grade assigned
# 90-100           O
# 80-89            A+
# 70-79            A
# 60-69            B
# 50-59            C
# 40-49            P
# Below 39         F

marks = int(input("Enter marks: "))

if marks > 100 or marks < 0:
    print("Invalid marks!")
elif marks >= 90:
    print("Grade O")
elif marks >= 80:
    print("Grade A+")
elif marks >= 70:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
elif marks >= 40:
    print("Grade P")
else:
    print("Grade F")
```
