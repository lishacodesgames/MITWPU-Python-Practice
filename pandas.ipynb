```python
import pandas as pd
a = [1, 2, 3]
```

```python
# Creating a series from list
pd.Series(a)
```

```python
# Create labels by changing index arguments
pd.Series(a, index=["a", "b", "c"])
```

```python
# Named indexes of dataframes

data = {
    "calories": [420, 300, 350],
    "duration": [50, 40, 45]
}

pd.DataFrame(data, index = ["Day 1", "Day 2", "Day 3"])
```

```python
# Data preprocessing: drop NaN
data = {
    'A': [1, 2, 3, None],
    'B': [None, None, 30, 40]
}

df = pd.DataFrame(data)
print("Original Data:\n", df, end='\n\n')

clean_df = df.dropna()
print("Clean Data:\n", clean_df)
```

```python
# Data preprocessing: replace None with some other value, instead of dropping the entire row
print("Original Data:\n", df, end='\n\n')

# inplace: False = returns a cleaned copy of df. True = changes original df, returns None
df.fillna(0, inplace=True)
print("Clean Data:\n", df)
```

```python
# Data preprocessing: use aggregates to fill missing values
data = {
    'A': [1, 2, 3, None],
    'B': [None, None, 30, 40]
}
df = pd.DataFrame(data)
print("Original Data:\n", df, end='\n\n')

df.fillna(df.mean(), inplace=True)
print("Clean Data:\n", df)
```
