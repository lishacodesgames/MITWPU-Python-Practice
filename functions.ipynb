```python
def get_parity(num: int):
    if(num % 2 == 0):
        return "Even"
    else:
        return "Odd"

print(get_parity(16))
print(get_parity(15))
```

```python
def sum(x, y, z):
    return x + y + z

print(sum(1, 2, 3))
```

```python
def sum_prod(x, y, z):
    s = x + y + z
    p = x * y * z
    return s, p

print(sum_prod(1, 3, 5))
```

```python
# display prime numbers of a number from user
def get_factors(num: int):
    factors = []
    for i in range(1, 1 + num):
        if num % i == 0:
            factors.append(i)

    return tuple(factors)

num = int(input("Enter a number: "))
print(get_factors(num))
```

```python
# print all even numbers from 1-100
def print_evens(end):
    for i in range(1, end + 1):
        print(i, end=" ")
```
