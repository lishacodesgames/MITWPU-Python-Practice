```python
# print the 4th element and 4th-last element

numbers = (10, 20, 30, 40, 50, 60, 70, 80, 90)
print("Tuple: ", numbers)
print("4th element: ", numbers[3])
print("4th-last element: ", numbers[-4])
```

```python
# check whether an element exists in tuple or not

element = int(input("Enter element to search: "))
if element in numbers:
    print("Element exists")
else:
    print("Element does not exist")
```

```python
# convert list to tuple

nums_list = [2, 35, 1, 6, 22]
nums_tuple = tuple(nums_list)
print("List: ", nums_list)
print("Tuple: ", nums_tuple)
```

```python
# find index of item in tuple

element = int(input("Enter element to find index of: "))
if element in numbers:
    print("Index = ", numbers.index(element))
else:
    print("Element not found")
```

```python
# replace last values of tuples in a list to 100
tuple_list = [(50, 50), (20, 5), (100, 50)]
print("List of tuples: ", tuple_list)

new_list = []
for tup in tuple_list:
    new_tup = tup[:-1] + (100,)
    new_list.append(new_tup)

print("After replacing last elm of tuple with '100': ", new_list)
```

```python
# find max and min elm of tuple

print("Tuple = ", nums_tuple)
print("max elm = ", max(nums_tuple))
print("min elm = ", min(nums_tuple))
```

```python
# swap first and last element of tuple
new_tuple = (numbers[-1],) + numbers[1:-1] + (numbers[0],)
print("Original: ", numbers)
print("After swapping: ", new_tuple)
```
