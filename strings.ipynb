```python
string = "Hello, World!"
print(string.lower())
print(string.upper(), end="\n\n")

for s in string:
    print(s)
```

```python
def splicer(string):
    if len(string) % 2 == 0:
        return 'even'
    else:
        return string[0]

print(splicer("hello!"))
```

```python
names = ["Lisha", "Mahak", "Surabhi", "Vivian"]
```

```python
list(map(splicer, names))
```
