```python
# Accept 20 values from user and save it in list. Perform following operations on it
#    a) count similar elements of list and print their index value   
#    b) count even and odd values of list 
#    c) count positive and negative values of list

while True:
    mylist = list(map(int, input("Enter 20 values: ").split()))
    if(len(mylist) != 20):
        print("20 values!")
    else:
        break
```

```python
# a)
print("(a)")
already_counted = []
for current in mylist:
    if current in already_counted:
        continue

    indexes = []
    for i in range(len(mylist)):
        if current == mylist[i]:
            indexes.append(i)

    already_counted.append(current)
    count = len(indexes)
    print(f"Element {current} appears {count} time%s" % ('s' if count > 1 else ''))
```

```python
# b)
print("(b)")
even_count, odd_count = 0, 0
for elm in mylist:
    if elm % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print("Even values: ", even_count)
print("Odd values: ", odd_count)
```

```python
# c)
print("(c)")
pos_count, neg_count = 0, 0
for elm in mylist:
    if elm > 0:
        pos_count += 1
    else:
        neg_count += 1

print("+ve values: ", pos_count)
print("-ve values: ", neg_count)
```

```python
# 10 Accept 10 values from user and save it in list. Perform following operations on it  
#   a) Sort list in ascending order using sorted() function and display sorted list
#   b) sort list in descending order using sort() function
#   c) display length of list

while True:
    mylist = list(map(int, input("Enter 10 values: ").split()))
    if(len(mylist) != 10):
        print("10 values!")
    else:
        break

print("List: ", mylist)
print("Ascending list: ", sorted(mylist))
mylist.sort(reverse=True)
print("Descending list: ", mylist)
```

```python
# Write a program that allows the user to type in a phrase and then outputs the acronym for that phrase. Note: the acronym should be all uppercase, even if the words in the phrase are not capitalized
phrase = list(input("Enter a phrase: ").split())

acronym = ""
for word in phrase:
    acronym += word[0].upper()

print(f"Acronym for this phrase is '{acronym}'")
```

```python
# print the abbreviation of a month, given its number
num = int(input("Enter a month's number: "))

match num:
    case 1:
        print("Jan")
    case 2:
        print("Feb")
    case 3:
        print("Mar")
    case 4:
        print("Apr")
    case 5:
        print("May")
    case 6:
        print("Jun")
    case 7:
        print("July")
    case 8:
        print("Aug")
    case 9:
        print("Sep")
    case 10:
        print("Oct")
    case 11:
        print("Nov")
    case 12:
        print("Dec")
    case _:
        print("That's not a month!")
```

```python
# merge two lists into a dictionary (one list as keys, another as values)
keys = ['apple', 'banana', 'cherry']
values = [1, 2, 3]

print("Keys: ", keys)
print("Values: ", values)

result = dict(zip(keys, values))
print("Dictionary: ", result)
```

```python
# remove duplicate elements from a list

numbers = [3, 1, 2, 3, 4, 1, 2]
unique_numbers = []

for item in numbers:
    if item not in unique_numbers:
        unique_numbers.append(item)

print(unique_numbers)
```

```python
# convert two lists into a dictionary using zip()

list1 = ["g", 12, 34.12]
list2 = [1, 2, 3]

mydict = dict(zip(list1, list2))
print("Dictionary: ", mydict)
```

```python
# find common elements in two lists

list1 = [1, 2, 2, 3, 4]
list2 = [2, 2, 4, 5, 6]

common_elements = [item for item in list1 if item in list2]
print(common_elements)
```
