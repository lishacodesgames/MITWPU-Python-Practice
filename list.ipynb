```python
mylist = [1, 2, 4, 4, 5]
print("My list: ", mylist)
```

```python
# create a list of 5 integers and display the list items. Access individual elements through index
print("Elements:", end=' ')
for i in range(len(mylist)):
    print(mylist[i], end=' ')
```

```python
# append a new item to the end of the list
mylist.append(6)
print("After appending: ", mylist)
```

```python
# reverse the order of the items in the list
mylist.reverse()
print("After reversing: ", mylist)
```

```python
# print the number of occurrences of a specified element in a list
print(f"4 appears {mylist.count(4)} times in my list")
```

```python
# append the items of list1 to list2 in the front
list1 = [1, 2, 3, 4]
list2 = [5, 6]

print(list1)
print(list2)

list2 = list1 + list2

print('-------')
print(list2)
```

```python
# insert a new item before the second element in an existing list

mylist.insert(1, 100)
print("After inserting: ", mylist)
```

```python
# remove the item from a specified index in a list

index = int(input("Enter index to remove: "))
if index < len(mylist):
    mylist.pop(index)
    print("After removing: ", mylist)
else:
    print("Index is too large!")
```

```python
# remove the first occurrence of a specified element from a list

element = int(input("Enter element to remove: "))
if element in mylist:
    mylist.remove(element)
    print("After removing: ", mylist)
else:
    print("Element not found")
```
